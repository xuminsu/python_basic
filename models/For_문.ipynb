### For 문

```python
# For 문 예제

test_list = ["one", "two", "three"]
for i in test_list:
    print(i)
```

```python
# 다양한 for 문 예제

a = [(1, 2), (3, 4), (5, 6)]
for (first, last) in a:
    print(first+last)
```

```python
# for문의 응용 문제

#총 5명의 학생이 시험을 보았는데, 60점 이상 합격, 그렇지 않으며 ㄴ불합격
student_dict = {"갑" : 70, "을" : 80, "병" : 45, "정" : 25, "무" : 90}

for name, score in student_dict.items():
    print(f"{name}: {'합격' if score >= 60 else '불합격'}")
```

```python
marks = [90, 25, 67, 45, 80]

number = 9
for mark in marks:
    number += 1
    if mark >= 60:
        print(f'{number}번 학생은 합격입니다.')
    else:
        print(f'{number}번 학생은 불합격입니다.')
```

```python
for num in range(len(marks)):   # range(5)
    if marks[num] < 60:      # 점수가 60점 미만인지 확인
        continue                # 처음으로 돌아가기 
    print(f'{num + 1}번 합격')
```

```python
# for와 range를 이용한 구구단

for i in range(2, 10):
    print("---" * 22)
    for j in range(1, 10):
        print(i * j, end = "\t")
    print("")
print("---" * 22)
```

```python
# 리스트 컴프리헨션(listj comprehension)  사용하기

a = [1, 2, 3, 4]
result = []
for num in a:
    result.append(num * 3)

print(result)
```

```python
a = [1, 2, 3, 4]
result = [num * 3 for num in a]
result
```

```python
# 만약 짝수에만 3을 곱하고 싶다면

a = [1, 2, 3, 4]
result = [num * 3 for num in a if num % 2 == 0] 
print(result)
```

```python
# for 문을 2개 이상 사용

result = [x * y for x in range(2, 10) for y in range(1, 10)]
print(result)
```

```python
# for 문과 break 문 (break 문은 for 문을 강제로 빠져나갈 때 사용)

for i in range(10):
    if i == 5:
        break
    print(i)
```

```python
#for - else 문

for i in range(5):
    print(i)
else:
    print("for 문이 정상 종료되었습니다.")
```

```python
# 만약 break로 빠져나가면 else 절은 출력되지 않는다.

for i in range(5):
    if i == 3:
        break
    print(i)
else : 
    print("for 문이 정상 종료되었습니다.")
```

```python
# enumerate 함수 활용하기 (리스트 순섯값 알고 싶을 때)

fruits = ["apple", "banana", "orange"]
for i, fruit in enumerate(fruits):
    print(f'{i}: {fruit}')
```

```python
# enumerate 함수 활용하기 (리스트 순섯값 알고 싶을 때)

fruits = ["apple", "banana", "orange"]
for i, fruit in enumerate(fruits, 1):
    print(f'{i}: {fruit}')
```

```python
# zip 함수로 여러 리스트 함께 순회하기
names = ["홍길동", "김철수", "이영희"]
scores = [85, 92, 78]
for name, score in zip(names, scores):
    print(f"{name}: {score}")
```

```python
# 세 개(이상) 의 리스트 순회
names = ["홍길동", "김철수", "이영희"]
kor = [85, 92, 78]
eng = [90, 88, 95]
for n, k, e in zip(names, kor, eng):
    print(f"{n}: 국어점수는 {k}점, 영어점수는 {e}점")

# kor, names, eng의 각 셀 일치해야 세개 다 나옴.
```

리스트 `words`에 ['apple'...] 가 있을 때 단어길이 출력

```python
words = ['apple', 'banana', 'cherry']

for i in range(len(words)):
    print(len(words[i]))

for word in words:
    print(len(word))
```
