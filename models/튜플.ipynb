변수 `a` 에 정수 10을, 변수 `b`에 실수 3.5를 저장하고 두 수의 합과 곱을 추력하는 코드를 작성하시오

```python
a = 10
b = 3.5
print(a + b)
print(a * b)
```

```python
# 복습차... f...
a, b = 10, 3.5
print(f"{a + b = }, {a * b = }")
```

변수 x 에 7, `y`에 3을 저장하고, 나늇셈 결과의 몫과 나머지를 출력하는 코드 작성하기

```python
x = 7; y = 3
print(x // y); print(x % y)

#다른방법...?
q = 0
r = x
while r >= y:     
    r -= y
    q += 1
print(q, r)        # 2 1
```
