변수 x 에 7, `y`에 3을 저장하고, 나늇셈 결과의 몫과 나머지를 출력하는 코드 작성하기

```python
x = 7; y = 3
print(x // y); print(x % y)

#다른방법...?
q = 0
r = x
while r >= y:     
    r -= y
    q += 1
print(q, r)        # 2 1
```

리스트 `numbers`에 [1, 2, 3, 4, 5]를 저장하고, 리스트의 첫 번쨰 요소와 마지막 요소를 출력하시오

```python
numbers = [1, 2, 3, 4, 5]
print(numbers[0], numbers[4])

for i in numbers[:]:
    if 1< i <5:
        numbers.remove(i)
print(numbers)
```
