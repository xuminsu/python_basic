## 파이썬 논리연산자

### 1. 기본 연산자
| 연산자 | 의미 | 예시 | 결과 |
|:---:|---|---|:---:|
| `and` | 둘 다 참이면 참 | `True and False` | `False` |
| `or` | 하나라도 참이면 참 | `True or False` | `True` |
| `not` | 참/거짓 뒤집기 | `not True` | `False` |

### 2. 진리표
| A | B | `A and B` | `A or B` | `not A` |
|:---:|:---:|:---:|:---:|:---:|
| True | True | True | True | False |
| True | False | False | True | False |
| False | True | False | True | True |
| False | False | False | False | True |

### 3. 우선순위
`not` → `and` → `or` 순서로 계산한다. 헷갈리면 괄호를 쓴다.
```python
True or False and False     # True  (and 먼저 → True or False)
(True or False) and False   # False
```

### 4. 거짓으로 취급되는 값 (Falsy)
`False`, `None`, `0`, `0.0`, `''`, `[]`, `()`, `{}`, `set()` → 나머지는 모두 참(Truthy)
```python
if []:          # 빈 리스트 → 거짓이라 실행 안 됨
    print("실행 안 됨")
```

### 5. 단락 평가 (Short-circuit)
결과가 정해지면 뒤쪽은 계산하지 않는다. 그리고 **True/False가 아니라 피연산자 값 자체를 돌려준다.**
| 식 | 결과 | 이유 |
|---|---|---|
| `0 and 10` | `0` | 앞이 거짓 → 앞 값 반환 |
| `5 and 10` | `10` | 앞이 참 → 뒤 값 반환 |
| `0 or "기본값"` | `'기본값'` | 앞이 거짓 → 뒤 값 반환 |
| `"이름" or "기본값"` | `'이름'` | 앞이 참 → 앞 값 반환 |

```python
name = input_name or "익명"     # 비어 있으면 기본값
x != 0 and 10 / x > 1          # x가 0이면 나눗셈을 안 해서 에러가 안 남
```

### 6. 비교 연결
```python
1 < x < 5          # (1 < x) and (x < 5) 와 같음
```

### 7. 함께 쓰는 멤버십·동일성 연산자
| 연산자 | 의미 | 예시 |
|:---:|---|---|
| `in` / `not in` | 포함 여부 | `'a' in 'apple'` → True |
| `is` / `is not` | 같은 객체인지 (`id` 비교) | `x is None` |
| `==` / `!=` | 값이 같은지 | `[1] == [1]` → True |

### 8. ⚠️ pandas·numpy에서는 `&`, `|`, `~`
여러 값을 한꺼번에 비교할 때는 `and`/`or`/`not` 대신 **비트 연산자**를 쓰고, **조건마다 괄호**를 친다.

| 파이썬 기본 | pandas / numpy |
|:---:|:---:|
| `and` | `&` |
| `or` | `\|` |
| `not` | `~` |

```python
df[(df['price'] > 12000) & (df['airline'] == 'Air_India')]   # O
df[(df['a'] > 1) | (df['b'] < 0)]                            # O
df[~df['col'].isna()]                                        # 결측 아닌 행
df[df['price'] > 12000 and df['airline'] == 'Air_India']     # X → ValueError
```
> `and`를 쓰면 `ValueError: The truth value of a Series is ambiguous` 에러가 난다.
> 괄호를 빼면 `&`가 비교 연산보다 먼저 계산돼서 결과가 틀리거나 에러가 난다.
