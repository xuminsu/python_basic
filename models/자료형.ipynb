````python
# 문자열이 숫자로 구성되어 있는지 확인하기 - isnumeric(), isdigit()
n = "123343"
n.isnumeric() 
n.isdigit()
````

보통 숫자(`"123343"`)에서는 둘 다 `True`라서 결과가 같아요. 차이는 **어떤 특수 문자까지 숫자로 인정하느냐**예요. 범위가 넓은 순서로 보면 `isnumeric()` ⊃ `isdigit()` ⊃ `isdecimal()`이에요.

| 문자열 | `isdecimal()` | `isdigit()` | `isnumeric()` |
|---|---|---|---|
| `'123'` | True | True | True |
| `'²'` (윗첨자) | False | True | True |
| `'①'` (원 숫자) | False | True | True |
| `'½'` (분수) | False | False | True |
| `'三'`, `'百'` (한자 숫자) | False | False | True |
| `'-5'` (음수) | False | False | False |
| `'3.14'` (소수) | False | False | False |
| `''` (빈 문자열) | False | False | False |

- `isdigit()`: 0~9와 윗첨자·원 숫자처럼 **숫자 모양 문자**까지 인정해요.
- `isnumeric()`: 거기에 분수, 한자 숫자처럼 **숫자 의미가 있는 문자**까지 인정해요.

**데이터 분석에서 알아 둘 점**

셋 다 `-`나 `.`이 들어가면 `False`예요. 그래서 `"-5"`나 `"3.14"` 같은 값은 숫자로 판단하지 못해요. 컬럼에 섞인 문자열을 숫자로 바꿀 때는 `pd.to_numeric`이 더 실용적이에요.

```python
df['col'] = pd.to_numeric(df['col'], errors='coerce')   # 숫자로 못 바꾸는 값은 NaN
df['col'].isna().sum()                                  # 몇 개가 숫자가 아니었는지
```

Series에 바로 쓰려면 `.str`을 붙이면 돼요.
```python
df['col'].str.isdigit()
```
