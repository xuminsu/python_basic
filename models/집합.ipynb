```python
# 교집합, 합집합, 차집합
s1 = set([1, 2, 3, 4, 5, 6])
s2 = set([4, 5, 6, 7, 8, 9])
```

```python
# 교집합
s1 & s2
```

```python
s1.intersection(s2)
```

```python
s1 | s2
```

```python
s1.union(s2)
```

```python
s1 - s2
```

```python
s2 - s1
```

```python
s1.difference(s2) # s1 - s2
```

```python
s2.difference(s1)
```

```python
# 값 1개 추가하기 - add

s1 = {1, 2, 3}
s1.add(4)
s1
```

```python
# 값 여러개 추가하기 - update
s1.update([5, 6, 7])
s1
```

```python
# 특정 값 제거하기 - remove
s1.remove(6)
s1
```

```python
# 특정 값 제거하기 - discard
s1.discard(7)
s1 # 없는 값도 오류가 나지 않음...
```

```python
s1.discard(8) 
s1 # 그 예시 
```

```python
s1.clear()
```

집합 `a`에 {1, 2, 3}, 'b' 에 {2, 3, 4}를 저장하고 두 집합의 교 / 합집합을 출력하시오 

```python
a = {1, 2, 3}
b = {2, 3, 4}
c = a - b
d = a & b
e = d - c
print(f"교집합은{e}이며, 합집합은{a | b} 이다") # e = a & b
```
