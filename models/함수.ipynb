### 함수

```python
# 함수 예시

def add(a, b):
    return a + b
```

```python
# 함수 사용

a = 10
b = 5
add(a, b)
```

```python
# 매개변수 / 인수
# 매개변수(Parameter) : 함수에 입력으로 전달된 값을 받는 변수
# 인수(Arguments) : 함수를 호출할 때 전달되는 입력 값

def add(a, b):          # a, b 매개변수
    return a + b

print(add(3, 4))        # 3, 4 는 인수
```

```python
# 일반적인 함수 

def add(a, b):
    result = a + b
    return result
```

```python
a = add(3, 4)
a
```

```python
# 입력값이 없는 함수

def say():
    return "Hi"
```

```python
a = say()
print(a)
```

```python
#반환값이 없는 함수 (return 문이 없다.)

def add(a, b):
    print(f"{a}, {b}의 합은 {a + b} 입니다.")
```

```python
add(3, 4)
```

```python
# 입력값도, 반환값도 없는 함수
def say():
    print("Hi")
```

```python
say()
```

```python
# 매개변수를 지정하여 호출하기
def sub(a, b):
    return b - a
```

```python
result = sub(a = 5, b = 3)
print(result)
```
